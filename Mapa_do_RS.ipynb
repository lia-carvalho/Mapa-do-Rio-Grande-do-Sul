```python
def busca_em_largura(grafo,origem,destino):
  fronteira = [(origem, [origem], 0)]  #Fila (cidade atual, caminho percorrido, custo acumulado )
  visitados = set()
  nos_expandidos = 0   #Para avaliar Complexidade de Tempo
  maximo_fronteira = 0 #Para avaliar Complexidade de Espaço

  while fronteira:
      if len(fronteira) > maximo_fronteira:
        maximo_fronteira = len(fronteira)

      no_atual,caminho,custo_atual = fronteira.pop(0)

      if no_atual in visitados:  #ignora se o nó atual já foi visitado
        continue

      visitados.add(no_atual)
      nos_expandidos += 1

      if no_atual == destino:
        return caminho, custo_atual, nos_expandidos, maximo_fronteira

      for vizinho, custoAresta in grafo[no_atual].items():
        if vizinho not in visitados:
          novo_caminho = caminho + [vizinho]
          novo_custo = custo_atual + custoAresta
          fronteira.append((vizinho, novo_caminho,novo_custo))

  return None, 0, nos_expandidos,maximo_fronteira

caminho_bfs , custo_bfs, expandidos_bfs, memoria_bfs = busca_em_largura(grafo, origem,destino)


print("---RESULTADOS BUSCA EM LARGURA---")
print("Caminho encontrado:",caminho_bfs)
print("Custo total (Otimalidade):", custo_bfs)
print("Nós expandidos(Tempo):",expandidos_bfs)
print("Máximo na fronteira(Espaço):", memoria_bfs)

```

```python
def busca_em_profundidade_sem_ciclo(grafo,origem,destino,limite_nos):
  fronteira = [(origem, [origem], 0)]  #Pilha (cidade atual, caminho percorrido, custo acumulado )
  nos_expandidos = 0   #Para avaliar Complexidade de Tempo
  maximo_fronteira = 0 #Para avaliar Complexidade de Espaço

  while fronteira:
      if len(fronteira) > maximo_fronteira:
        maximo_fronteira = len(fronteira)

      no_atual,caminho,custo_atual = fronteira.pop()  #remove o último da pilha
      nos_expandidos += 1

      if no_atual == destino:
        return caminho, custo_atual, nos_expandidos, maximo_fronteira

      if nos_expandidos >= limite_nos:  #interrompe provável loop infinito
        return None, 0, nos_expandidos, maximo_fronteira

      for vizinho, custoAresta in grafo[no_atual].items():  #não verifica se já foi visitado
        novo_caminho = caminho + [vizinho]
        novo_custo = custo_atual + custoAresta
        fronteira.append((vizinho, novo_caminho,novo_custo))

  return None, 0, nos_expandidos,maximo_fronteira

limite = 1000  #limite de nos ampliado para evitar travar o código

print("---RESULTADOS DFS SEM DETECÇÃO DE CICLOS---")
for origem_teste in origens:
  caminho_dfs, custo_dfs, expandidos_dfs, memoria_dfs = busca_em_profundidade_sem_ciclo(grafo, origem_teste, destino, limite)
  print("\nOrigem:", origem_teste)
  if caminho_dfs is None:
    print("Limite de", limite, "nós atingido: loop infinito (sem solução)")
  else:
    print("Caminho encontrado:", caminho_dfs)
    print("Custo total (Otimalidade):", custo_dfs)
  print("Nós expandidos(Tempo):", expandidos_dfs)
  print("Máximo na fronteira(Espaço):", memoria_dfs)
```

```python
def busca_best_first(grafo, heuristica, origem, destino):
  contador = 0  #desempate na fila de prioridade (ordem de inserção)
  fronteira = [(heuristica[origem], contador, origem, [origem], 0)]  #Fila de prioridade (h(n), desempate, cidade atual, caminho percorrido, custo acumulado)
  visitados = set()
  nos_expandidos = 0   #Para avaliar Complexidade de Tempo
  maximo_fronteira = 0 #Para avaliar Complexidade de Espaço

  while fronteira:
      if len(fronteira) > maximo_fronteira:
        maximo_fronteira = len(fronteira)

      h_atual, _, no_atual, caminho, custo_atual = heapq.heappop(fronteira)  #remove o nó com MENOR h(n)

      if no_atual in visitados:  #ignora se o nó atual já foi visitado
        continue

      visitados.add(no_atual)
      nos_expandidos += 1

      if no_atual == destino:
        return caminho, custo_atual, nos_expandidos, maximo_fronteira

      for vizinho, custoAresta in grafo[no_atual].items():
        if vizinho not in visitados:
          contador += 1
          novo_caminho = caminho + [vizinho]
          novo_custo = custo_atual + custoAresta
          heapq.heappush(fronteira, (heuristica[vizinho], contador, vizinho, novo_caminho, novo_custo))

  return None, 0, nos_expandidos, maximo_fronteira

print("--- RESULTADOS BEST-FIRST COM DETECÇÃO DE CICLOS ---")
for origem_teste in origens:
  caminho_bf, custo_bf, expandidos_bf, memoria_bf = busca_best_first(grafo, heuristica, origem_teste, destino)
  print("\nOrigem:", origem_teste)
  print("Caminho encontrado:", caminho_bf)
  print("Custo total (Otimalidade):", custo_bf)
  print("Nós expandidos(Tempo):", expandidos_bf)
  print("Máximo na fronteira(Espaço):", memoria_bf)
```

```python
def busca_ids(grafo, origem, destino):
  total_expandidos = 0   #soma de TODAS as iterações (o IDS re-expande nós)
  maximo_fronteira = 0   #maior pilha vista em qualquer iteração

  for limite_prof in range(len(grafo)):  #um caminho simples tem no máximo len(grafo)-1 arestas
    fronteira = [(origem, [origem], 0)]  #Pilha (cidade atual, caminho percorrido, custo acumulado)
    houve_corte = False  #True se algum nó deixou de ser expandido por causa do limite

    while fronteira:
        if len(fronteira) > maximo_fronteira:
          maximo_fronteira = len(fronteira)

        no_atual, caminho, custo_atual = fronteira.pop()  #remove o último da pilha (LIFO)
        total_expandidos += 1

        if no_atual == destino:
          return caminho, custo_atual, total_expandidos, maximo_fronteira

        if len(caminho) - 1 >= limite_prof:  #atingiu a profundidade máxima desta iteração
          houve_corte = True
          continue

        for vizinho, custoAresta in grafo[no_atual].items():
          if vizinho not in caminho:  #evita ciclo no caminho atual
            novo_caminho = caminho + [vizinho]
            novo_custo = custo_atual + custoAresta
            fronteira.append((vizinho, novo_caminho, novo_custo))

    if not houve_corte:  #árvore inteira explorada sem cortes: não existe solução
      break

  return None, 0, total_expandidos, maximo_fronteira

print("\n--- RESULTADOS IDS COM DETECÇÃO DE CICLOS ---")
for origem_teste in origens:
  caminho_ids, custo_ids, expandidos_ids, memoria_ids = busca_ids(grafo, origem_teste, destino)
  print("\nOrigem:", origem_teste)
  print("Caminho encontrado:", caminho_ids)
  print("Custo total (Otimalidade):", custo_ids)
  print("Nós expandidos(Tempo):", expandidos_ids)
  print("Máximo na fronteira(Espaço):", memoria_ids)
```

```python
def busca_custo_uniforme(grafo, origem, destino):
    contador = 0  # Desempate na fila de prioridade (ordem de inserção)
    # Fila de prioridade: (custo_acumulado g(n), desempate, cidade_atual, caminho_percorrido)
    fronteira = [(0, contador, origem, [origem])]
    visitados = set()
    nos_expandidos = 0   # Para avaliar Complexidade de Tempo
    maximo_fronteira = 0 # Para avaliar Complexidade de Espaço

    while fronteira:
        if len(fronteira) > maximo_fronteira:
            maximo_fronteira = len(fronteira)

        custo_atual, _, no_atual, caminho = heapq.heappop(fronteira)  # Remove o nó com MENOR g(n)

        if no_atual in visitados:  # Ignora se o nó atual já foi visitado
            continue

        visitados.add(no_atual)
        nos_expandidos += 1

        if no_atual == destino:
            return caminho, custo_atual, nos_expandidos, maximo_fronteira

        for vizinho, custo_aresta in grafo[no_atual].items():
            if vizinho not in visitados:
                contador += 1
                novo_caminho = caminho + [vizinho]
                novo_custo = custo_atual + custo_aresta
                heapq.heappush(fronteira, (novo_custo, contador, vizinho, novo_caminho))

    return None, 0, nos_expandidos, maximo_fronteira


# Execução para as 3 origens obrigatórias
origens = ['Santa Vitoria do Palmar', 'Erexim', 'Uruguaiana']
destino = 'Porto Alegre'

print("--- RESULTADOS BUSCA DE CUSTO UNIFORME (UCS) ---")
for origem_teste in origens:
    caminho_ucs, custo_ucs, expandidos_ucs, memoria_ucs = busca_custo_uniforme(grafo, origem_teste, destino)
    print("\nOrigem:", origem_teste)
    print("Caminho encontrado:", caminho_ucs)
    print("Custo total (Otimalidade):", custo_ucs)
    print("Nós expandidos (Tempo):", expandidos_ucs)
    print("Máximo na fronteira (Espaço):", memoria_ucs)
```
